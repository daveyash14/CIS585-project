```python
from PIL import Image
import os

def resize_images(folder_path):
    for filename in os.listdir(folder_path):
        if filename.endswith(('.jpg', '.jpeg', '.png', '.bmp')):  # Add more extensions if needed
            img_path = os.path.join(folder_path, filename)
            with Image.open(img_path) as img:
                width, height = img.size
                new_width = int(width / 2)
                new_height = int(height / 2)
                resized_img = img.resize((new_width, new_height))
                resized_img.save(img_path)  # Overwrite the original image

folder_path = '/Users/yashdave/Documents/MSc/CIS583/explaining_ai/StableDiffusion-PyTorch/data/CelebAMask-HQ/CelebA-HQ-img'
resize_images(folder_path)
```
