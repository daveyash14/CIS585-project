```python
import os

caption_path = "dataset_full/celeba-caption"
image_path = "dataset_full/CelebA-HQ-img"

num_to_transfer = 10

cap_dest_path = "celeba-caption"
img_dest_path = "CelebA-HQ-img"

def transfer_files(caption_path, image_path, cap_dest_path, img_dest_path, num_to_transfer):
    # Create destination directories if they don't exist
    os.makedirs(cap_dest_path, exist_ok=True)
    os.makedirs(img_dest_path, exist_ok=True)

    # Get the list of files in the caption and image directories
    caption_files = sorted(os.listdir(caption_path))
    image_files = sorted(os.listdir(image_path))

    # Check if the number of files is sufficient
    if len(caption_files) < num_to_transfer or len(image_files) < num_to_transfer:
        raise ValueError("Not enough files to transfer.")

    # Transfer the specified number of files
    for i in range(num_to_transfer):
        cap_file = caption_files[i]
        img_file = image_files[i]

        # Copy the caption file
        os.system(f"cp {os.path.join(caption_path, cap_file)} {cap_dest_path}")

        # Copy the image file
        os.system(f"cp {os.path.join(image_path, img_file)} {img_dest_path}")
```

```python
transfer_files(caption_path, image_path, cap_dest_path, img_dest_path, num_to_transfer)
```
